# Loops

Python loops: `for`, `while`, `break`, `else`, nested loops, and iteration patterns.

## For Loop with Range

```python
# Range
for i in range(5):
    print(i)
```

## For Loop with List

```python
# list
ages = [1, 2, 40, 12, 5]
for i in ages:
    print("age: ", i)
```

## While Loop

```python
b = -1
while b <= 0:
    arg = input("enter a positive value: ")
    b = int(arg)

print("Given value is: ", b)
```

## Pattern Printing

```python
for num in range(1, 5):
    print(num * "*")
```

## Range with Step

```python
for num in range(0, 10, 2):
    print(num)
```

## For & Break

```python
numbers = [3, 7, 9, 2, 5]

for n in numbers:
    if n == 2:
        print("Found 2 \u2014 stopping loop")
        break
    print("Checking:", n)
```

## For-Else

The `else` block executes only if the loop completes **without** hitting `break`.

```python
# 2 is NOT in the list - else will execute
numbers = [3, 7, 9, 5]

for n in numbers:
    if n == 2:
        print("Found 2")
        break
else:
    print("2 not found in the list")
```

```python
# 2 IS in the list - else will NOT execute
numbers = [3, 2, 9, 5]

for n in numbers:
    if n == 2:
        print("Found 2")
        break
else:
    print("2 not found in the list")
```

## While-Else

The `else` block runs only if the while loop finishes normally (NO `break` used).

```python
# Loop completes normally - else executes
x = 1

while x <= 5:
    print("x =", x)
    x += 1
else:
    print("Loop finished without break")
```

```python
# Loop breaks at x==3 - else will NOT execute
x = 1

while x <= 5:
    print("x =", x)
    if x == 3:
        break
    x += 1
else:
    print("Loop finished")
```

## Nested Loops

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(f"i = {i}, j = {j}")
```

## Range Type

```python
print(type(range(5)))  # <class 'range'> - it is iterable
```

## Iterating Over Strings

```python
text = "Python"

for ch in text:
    print(ch)
```

```python
# Using while loop
text = "Hello"
i = 0

while i < len(text):
    print(text[i])
    i += 1
```

## Iterating Over Lists

```python
fruits = ["apple", "banana", "mango"]

for item in fruits:
    print(item)
```

## Enumerate

`enumerate()` gives both index and value.

```python
text = "World"

for index, ch in enumerate(text):
    print(index, ch)
```

```python
# Set with enumerate
numbers = {10, 20, 30}

for index, value in enumerate(numbers):
    print(index, value)
```

## Dictionary Iteration

```python
person = {"name": "Ravi", "age": 25, "city": "Hyderabad"}

# Iterate keys
print("------key------")
for key in person:
    print(key)

# Iterate values
print("------value------")
for value in person.values():
    print(value)

# Iterate key & value
print("------key & value------")
for key, value in person.items():
    print(key, "=>", value)
```

```python
# Enumerate with dict items
person = {"name": "Ravi", "age": 25, "city": "Hyderabad"}

for index, (key, value) in enumerate(person.items()):
    print(index, key, value)
```
