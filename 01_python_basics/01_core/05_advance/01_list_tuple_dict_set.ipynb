# Lists, Tuples, Dictionaries & Sets

Python's core collection data structures.

## Overview

```python
lst = [10, 20, 30]
tpl = (1, 2, 3)  # fixed
dct = {"name": "Ravi", "age": 25}  # dictionary
st = {1, 2, 3, 3}  # set

lst.append(40)
print(lst, tpl, dct["name"], st)
```

## List

- Ordered, mutable, allows duplicates
- Can modify elements by index

```python
fruits = ["apple", "banana", "orange", "banana"]
fruits[1] = "grape"  # allowed
print(fruits)
```

## Tuple

- Ordered, **immutable**, allows duplicates
- Cannot modify elements after creation

```python
colors = ("red", "green", "blue", "green")
# colors[1] = "yellow"  # error (cannot modify)
print(colors)
```

## Set

- Unordered, no duplicates
- Fast membership testing with `in`

```python
numbers = {1, 2, 3, 3, 4}
print(numbers)  # output: {1, 2, 3, 4} (removes duplicate)
print(4 in numbers)  # True
```

## Dictionary

- Key-value pairs, ordered (Python 3.7+), mutable
- Keys must be unique

```python
person = {"name": "Ravi", "age": 25}
person["age"] = 26  # update value
print(person["name"])
```
