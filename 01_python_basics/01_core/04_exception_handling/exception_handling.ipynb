# Exception Handling

Using `try`, `except`, and `finally` to handle runtime errors gracefully.

## Basic try-except-finally

```python
try:
    num = int(input("Enter number: "))
    print("Square:", num * num)
except ValueError:
    print("Invalid number")
finally:
    print("Go to another")
```

## Catching Exception Details with `as`

```python
try:
    num = int(input("Enter number: "))
    print("Square:", num * num)
except ValueError as e:
    print("Invalid number")
    print("Error info:", e)
finally:
    print("Program done")
```
