# Hello World - Basic Python syntax

The simplest Python program starts with `print()`. No boilerplate code needed.

```python
print("Hello World")
print("Today is a good day")
```

## Variables and Arithmetic

No data type declaration needed in Python. Variables are dynamically typed.
Print can have multiple arguments separated by comma.

```python
# Single line comment
# no data type declaration needed
a = 2
b = 3.2
c = a + b

'''
Multiline comment
 print can have multiple args, sperated by comma
'''
print(a, "+", b, "=", c)
```

## User Input

`input()` always returns a string. Use `int()` or `float()` to convert.
f-strings allow embedding expressions inside `{}` in a string.

```python
x = input("Enter Input: ")  # treated as string
print(x)
x = int(x)
print(f"printing {x} stars") # formatted String
print("*"*x)
```
