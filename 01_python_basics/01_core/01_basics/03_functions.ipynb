# Functions

Functions are reusable blocks of code defined with the `def` keyword.

## Basic Function (no return)

A function without a `return` statement returns `None` by default.

```python
def greet():
    print("Hello, good morning")
    # this function returns 'None'


greet()
```

## Function with Return Value

Parameters are placeholders; arguments are the actual values passed.

```python
def sum(a, b): # a,b are parameters
    return a+b # return type


f = 2
s = 3
total = sum(f, s) # f,s are args

print(f, "+", s, "=", total)
print(greet()) # None
```

## Keyword Arguments

You can pass arguments by name for clarity and order independence.

```python
def wish(name, message):
    print(f"{message}, {name}")

wish(name="Ravi", message="Good Morning")
```

## Default Arguments

Default parameters must come after required parameters.

```python
# def increament(number, by=1, multiple): # invalid
def increament(number, by=1):
    return number + by

print(increament(3))
print(increament(3,2))
print(increament(3, by=3))
```

## Variable Arguments (*args)

`*args` allows a function to accept any number of positional arguments as a tuple.

```python
def add(*nums):
    total=0
    for n in nums:
        total+=n
    return total


print(add(1,2,3,4,5))
```
