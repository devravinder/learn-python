# Type Casting - Converting between types

Python provides built-in functions to convert between data types: `int()`, `float()`, `bool()`, `str()`.

## int() conversion

```python
arg = input("Enter a number: ")
x = int(arg)
print("int:", x, type(x))
```

## float() conversion

```python
arg = input("Enter a float: ")
x = float(arg)
print("float:", x, type(x))
```

## bool() conversion

```python
arg = input("Enter a bool: ")
x = bool(arg)
print("bool:", x , type(x))
```
