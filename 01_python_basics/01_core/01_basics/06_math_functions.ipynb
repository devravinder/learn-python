# Math Functions

Python provides built-in math functions and a `math` module for advanced operations.

## Built-in Math Functions

`round()` rounds to the nearest integer, `abs()` returns the absolute value.

```python
x=9.3
print(round(x)) # built-in
print(abs(x))
```

## Math Module

The `math` module provides functions like `ceil()`, `floor()`, `sqrt()`, etc.

```python
import math

print(math.ceil(x))
```
