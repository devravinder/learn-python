## Complex Types

Lists, dictionaries, and other reference types. Use `type()` to check a variable's type.

```python
name = "Ravi"
age = 25
height = 5.9
is_married = False
skills = ["Python", "JavaScript"] # list
person = {"name": name, "age": age} # dictionary

print(type(name), type(age), type(height), type(skills))
```
