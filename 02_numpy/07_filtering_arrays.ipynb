# Filtering Arrays

```python
import numpy as np

ages = np.array([[10, 19, 18, 28, 54, 65, 33], [1, 17, 27, 34, 87, 69, 77]])
print("ages", ages)
```

## Boolean Indexing

```python
# Teenagers (< 18)
teenagers = ages[ages < 18]
print(teenagers)
```

```python
# Adults (>= 18 and < 65)
adults = ages[(ages >= 18) & (ages < 65)]
print(adults)
```

```python
# Good people (< 18 or > 65)
good_people = ages[(ages < 18) | (ages > 65)]
print(good_people)
```

```python
# Seniors (> 65)
seniors = ages[ages > 65]
print(seniors)
```

```python
# Even ages (% 2 == 0)
evens = ages[ages % 2 == 0]
print(evens)
```

## np.where() - Preserves Dimensions

Unlike boolean indexing which flattens the result, `np.where()` preserves the original array shape.

```python
adults = np.where(ages > 18, ages, 0)
print(adults)
```
