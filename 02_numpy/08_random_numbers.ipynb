# Random Numbers in NumPy

## Random Integer Generation (default_rng)

```python
import numpy as np

rng = np.random.default_rng()
print(rng.integers(low=1, high=100))
print(rng.integers(low=1, high=100, size=3))
print(rng.integers(low=1, high=100, size=(3, 2)))
```

## Seed (Reproducible Results)

```python
rng = np.random.default_rng(seed=1)
print(rng.integers(low=1, high=100))
print(rng.integers(low=1, high=100, size=3))
print(rng.integers(low=1, high=100, size=(3, 2)))
```

## Uniform Distribution (old style)

```python
print(np.random.uniform())
print(np.random.uniform(low=-1, high=1))
print(np.random.uniform(low=-1, high=1, size=3))
print(np.random.uniform(low=-1, high=1, size=(3, 2)))
```

## Uniform with Seed

```python
np.random.seed(seed=1)
print(np.random.uniform(low=-1, high=1, size=(3, 2)))
```

## Shuffle & Choice

```python
rng = np.random.default_rng()
array = np.array([1, 2, 3, 4, 5])
rng.shuffle(array)
print(array)

fruits = np.array(["apple", "banana", "cherry", "grape", "mango"])
fruit = rng.choice(fruits)
print(fruit)
three_fruits = rng.choice(fruits, size=3)
print(three_fruits)
many_fruits = rng.choice(fruits, size=(3, 3))
print(many_fruits)
too_many_fruits = rng.choice(fruits, size=(3, 3, 3))
print(too_many_fruits)
```
